# Online Retail Data Cleaning

This notebook performs initial cleaning and preprocessing of the Online Retail dataset.

Key tasks:
- Inspect dataset structure
- Handle missing values
- Remove cancelled transactions
- Remove negative quantities
- Create revenue feature
- Save cleaned dataset for analysis

## Import Libraries and Load Dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\Datasets\Research\retail_analytics_sql_python\data\raw\Online_Retail.csv", encoding="ISO-8859-1")

df.head()
```

```python
# Dataset size
df.shape
```

```python
# Dataset Structure
df.info()
```

```python
# Summary Statistics
df.describe()
```

```python
# Missing Values
df.isna().sum().sort_values(ascending=False)
```

```python
# Remove Missing Customer IDs (Customer ID is required for customer-level analysis and segmentation)
df = df.dropna(subset=["CustomerID"])
```

```python
# Remove Cancelled Transactions (Invoices starting with "C" represent cancelled orders).
df = df[~df["InvoiceNo"].astype(str).str.startswith("C")]
```

```python
# Remove Negative Quantities
df = df[df["Quantity"] > 0]
```

```python
# Remove Invalid Prices
df = df[df["UnitPrice"] > 0]
```

```python
# Create Revenue Feature Using Quantity and Unit Price
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

```python
# Convert Invoice Date to Date Format (The dataset uses day-first datetime format (`dd/mm/yyyy hh:mm`), so `dayfirst=True` is required.)
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], dayfirst=True)
```

```python
# Save the Cleaned Dataset
df.to_csv("../data/processed/retail_cleaned.csv", index=False)
```
