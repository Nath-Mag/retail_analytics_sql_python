# Business Insights and Recommendations

This notebook summarizes key findings from the retail analytics project and provides actionable business recommendations.

The analysis focuses on:

- Revenue distribution
- Product performance
- Customer purchasing behavior
- Customer segmentation

```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Processed Data  
df = pd.read_csv("../data/processed/retail_cleaned.csv")

rfm = pd.read_csv("../data/processed/customer_rfm_segments.csv")
```

## Revenue by Country

Understanding which markets generate the most revenue helps prioritize regional strategies.

```python
country_revenue = df.groupby("Country")["Revenue"].sum().sort_values(ascending=False).head(10)

country_revenue.plot(
    kind="bar",
    title="Top Revenue Generating Countries"
)
plt.xticks(rotation=45)
plt.show()
```

## Top Products by Revenue

```python
top_products = df.groupby("Description")["Revenue"].sum().sort_values(ascending=False).head(10)

top_products.plot(
    kind="barh",
    title="Top Revenue Products"
)

plt.show()
```

## Monthly Revenue Trend

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df["Month"] = df["InvoiceDate"].dt.to_period("M")

monthly_revenue = df.groupby("Month")["Revenue"].sum()

monthly_revenue.plot(
    figsize=(10,5),
    title="Monthly Revenue Trend"
)

plt.show()
```

## Customer Segmentation Analysis

Customers were segmented using RFM analysis

```python
rfm["Segment"].value_counts().plot(
    kind="bar",
    title= "Customer Segment Distribution",
)
plt.xticks(rotation=360)
plt.show()
```

## Key Business Insights

### 1. Market Concentration

The revenue distribution by country shows that the **United Kingdom overwhelmingly dominates total revenue**, contributing the majority of sales compared to all other markets.

This indicates that the business currently relies heavily on the UK market, with relatively limited penetration in other European countries such as the Netherlands, Germany, France, and Ireland.

This concentration creates both an opportunity and a risk:
- Opportunity to expand internationally
- Risk due to dependence on a single market



### 2. Product Performance

The analysis of top revenue-generating products reveals that a relatively small number of products drive a large share of total revenue.

Notable high-performing items include:

- PAPER CRAFT , LITTLE BIRDIE
- REGENCY CAKESTAND 3 TIER
- WHITE HANGING HEART T-LIGHT HOLDER
- JUMBO BAG RED RETROSPOT
- MEDIUM CERAMIC TOP STORAGE JAR

These products appear to be decorative or gift-related items, suggesting strong demand for home décor and novelty products.

This indicates that **product categories related to gifts, decorations, and seasonal items are major revenue drivers**.



### 3. Seasonal Sales Patterns

The monthly revenue trend shows clear **seasonal purchasing behavior**.

Revenue increases significantly from **September through November**, peaking in **November** before dropping in December.

This pattern strongly suggests **holiday season purchasing**, where customers increase spending on gifts and decorations leading up to major holidays.

Understanding this seasonal pattern is critical for:
- inventory planning
- marketing campaigns
- staffing and logistics

## Strategic Business Recommendations

Based on the analysis, several strategic actions can improve business performance.

### 1. Customer Retention Strategy

The RFM segmentation shows that the majority of customers fall into the **Regular Customers** segment, while a smaller portion are classified as **Best Customers**.

Best customers contribute disproportionately to revenue and should be prioritized.

Recommended actions:

- Implement loyalty programs
- Offer exclusive promotions for high-value customers
- Provide early access to new products



### 2. Re-engage At-Risk Customers

Customers who previously purchased but have not returned recently represent a churn risk.

Targeted campaigns could include:

- personalized email promotions
- discount offers
- reminder campaigns for seasonal purchases

This can significantly increase repeat purchase rates.



### 3. Optimize Product Strategy

Since a small number of products drive most revenue:

- ensure consistent inventory for top-performing products
- increase marketing for high-margin items
- introduce complementary products within the same category

This strategy can increase average basket size.



### 4. Expand International Sales

Although the UK dominates sales, other European markets show demand potential.

The company could expand through:

- targeted international marketing
- localized shipping strategies
- partnerships with regional distributors

Diversifying geographically would reduce dependence on a single market.
